```python
import pandas as pd

df= pd.read_csv("Hanna Film Dataset.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
leny=df.sort_values(by="Year", ascending=True)
leny.plot(kind="bar", x="Year", y="Duration in minutes", legend=False)
plt.ylabel("Duration")
plt.title("Film length per year")
plt.show
```

This column chart shows the length of movies by the year they were released, showing that overall movies in the dataset have gotten a little bit shorter on average. The longest movie is also the oldest in the dataset. 

```python
#pie chart
avg_rating=df.groupby("Genre")["RT Rating"].mean().sort_values(ascending=False)
plt.figure(figsize=(8, 8))
plt.pie(avg_rating, labels=avg_rating.index, autopct='%1.2f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.show
```

This pie chart shows the genres' average ratings in relation to each other

```python
plt.scatter(df["RT Rating"], df["Year"], s=df["RT Rating"]*10, alpha=0.3)
plt.ylabel("Year")
plt.xlabel("RT Rating")
plt.title("Uhh")
plt.show
```
