```python
import pandas as pd

df= pd.read_csv("Hanna Film Dataset.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
leny=df.sort_values(by="Year", ascending=True)
leny.plot(kind="bar", x="Year", y="Duration in minutes", legend=False)
plt.ylabel("Duration")
plt.title("Film length per year")
plt.show
```

This column chart shows the length of movies by the year they were released, showing that overall movies in the dataset have gotten a little bit shorter on average. The longest movie is also the oldest in the dataset. 

```python
#pie chart
avg_rating=df.groupby("Genre")["RT Rating"].mean().sort_values(ascending=False)
plt.figure(figsize=(8, 8))
plt.pie(avg_rating, labels=avg_rating.index, autopct='%1.2f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.show
```

This pie chart shows the genres' average ratings in relation to each other

```python
plt.scatter(df["Duration in minutes"], df["RT Rating"], s=df["RT Rating"]*10, alpha=0.3)
plt.ylabel("Rating")
plt.xlabel("Duration")
plt.title("Rating vs film duration")
plt.show
```

This scatter plot shows the rating of films in relation to their duration. It shows that most of the higher rated films are shorter, and the longest film in the dataset has the lowest rating. However, the second longest film has a much higher rating.

```python
rt_min = df["RT Rating"].min()
rt_max = df["RT Rating"].max()
rt_q1 = df["RT Rating"].quantile(0.25)
rt_q3 = df["RT Rating"].quantile(0.75)
ln_min = df["Duration in minutes"].min()
ln_max = df["Duration in minutes"].max()
ln_q1=df["Duration in minutes"].quantile(0.25)
ln_q3=df["Duration in minutes"].quantile(0.75)
print("RT Rating - min:", rt_min, "Max:", rt_max, "Q1:", rt_q1, "Q3:", rt_q3)
print("Duration - min:", ln_min, "Max:", ln_max, "Q1:", ln_q1, "Q3:", ln_q3)
df[["RT Rating", "Duration in minutes"]].boxplot()
plt.show
```

The box plot shows the consistency of ratings and film lengths. The lowest rating is 82, and the highest is 99. 25% of the films have a rating below 91 and 75% above; 75% of the films are rated below 96 and 25% higher. The shortest movie is 98 minutes and the longest is 188 minutes (over 3 hours). 25% of the films are shorter than 116 minutes, and 75% are longer. 75% of films are shorter than 147 minutes (about 2,5 hours) and 25% are longer.
