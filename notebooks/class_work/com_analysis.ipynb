```python
import pandas as pd

df= pd.read_csv("comp.csv")
df.head()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)
plt.ylabel("Hours")
plt.title("Hours Studied per Student")
plt.show
```

```python
# et oleks nime average mitte sada nime korraga

import matplotlib.pyplot as plt
avg_hours=df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)
plt.ylabel("Hours")
plt.title("Hours Studied per Student")
plt.show
```

```python
#who performs best

import matplotlib.pyplot as plt
avg_grade=df.groupby("Student")["Grade"].mean().sort_values(ascending=False)
avg_grade.plot(kind="bar", x="Student", y="Grade", legend=False)
plt.ylabel("Grade")
plt.title("Avg grade of Student")
plt.show
```

```python
#pie chart
import matplotlib.pyplot as plt
avg_grade=df.groupby("Student")["Grade"].mean().sort_values(ascending=False)
plt.figure(figsize=(8, 8))
plt.pie(avg_grade, labels=avg_grade.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.show
#autopct - % of values 1.1 decimal place
# startangle rotates
# plt.cm diff colors for diff slices
# those things are optional
```

```python
# does more studying give better grades? let's find out! 
import matplotlib.pyplot as plt
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.3)
plt.ylabel("Grade")
plt.xlabel("Hours studied")
plt.title("Grade vs hours studied")
plt.show
```

```python
# which course & which program has highest avg performance
avg_g=df.groupby("Course")["Grade"].mean() #.sort_values(ascending=False)
print("Average Grade per Course:\n", avg_g, "\n")
avg_g.plot(kind="bar", color="orchid")
plt.ylabel("Grade")
plt.title("Avg grade of Course")
plt.show

#Programmiga sama
```

```python
#how consistent are study patterns and performance?
#plt.figure(figsize=8,5)
hr_min = df["Hours_Studied"].min()
hr_max = df["Hours_Studied"].max()
gr_min = df["Grade"].min()
gr_max = df["Grade"].max()
gr_q1=df["Grade"].quantile(0.25)
print("Hours_Studied - min:", hr_min, "Max:", hr_max)
print("Grades - min:", gr_min, "Max:", gr_max, "Q1:", gr_q1)
df[["Hours_Studied", "Grade"]].boxplot()
plt.show
```
